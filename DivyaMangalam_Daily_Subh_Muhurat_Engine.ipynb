```python
#@title 3. 🕉️ Configure Daily Video Parameters
#@markdown Choose the date, deity, and videos to generate.

target_date = "today" #@param ["today", "tomorrow"] {allow-input: true}
selected_deity = "auto" #@param ["auto", "surya", "shiva", "hanuman", "ganesha", "venkateswara", "durga", "ayyappa"]
video_bundle = "5min_shiva" #@param ["5min_shiva", "all", "panchang", "deity", "darshan", "5min_suprabhatam", "5min_ayyappa", "5min_durga", "5min_ramayana", "mahabharata_p1"]
publish_to_instagram = True #@param {type:"boolean"}
publish_to_youtube = True #@param {type:"boolean"}

import datetime
utc_now = datetime.datetime.now(datetime.timezone.utc)
ist_now = utc_now + datetime.timedelta(hours=5, minutes=30)

if target_date == "today":
    run_date_str = ist_now.strftime("%Y-%m-%d")
elif target_date == "tomorrow":
    run_date_str = (ist_now + datetime.timedelta(days=1)).strftime("%Y-%m-%d")
else:
    run_date_str = target_date

pub_flag = "all" if (publish_to_instagram and publish_to_youtube) else ("instagram" if publish_to_instagram else ("youtube" if publish_to_youtube else "none"))

print("===================================================")
print(f"📅 Target Date (IST):  {run_date_str}")
print(f"🕉️ Selected Deity:     {selected_deity.upper()}")
print(f"🎬 Video Bundle:       {video_bundle.upper()}")
print(f"🚀 Auto-Publish:       {pub_flag.upper()}")
print("===================================================")
```

```python
#@title 4. 🎬 Render Multi-Deity Subh Muhurat Videos & Auto-Publish
#@markdown Generates all high-definition videos with pure sacred music and anti-theft watermark.

!python cloud_daily_subh_muhurat_pipeline.py \
  --mode "{video_bundle}" \
  --deity "{selected_deity}" \
  --publish "{pub_flag}" \
  --date "{run_date_str}"

print("\n🎉 Video Generation and Multi-Platform Publishing Completed!")
```
