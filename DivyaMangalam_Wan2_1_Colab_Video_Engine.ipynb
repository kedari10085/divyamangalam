```
# @title 2. Load Wan 2.1 Video Foundation Model (Alibaba Open-Weights)
import torch
from diffusers import WanPipeline
from diffusers.utils import export_to_video

# Model options:
# 'Wan-AI/Wan2.1-T2V-1.3B-Diffusers' -> Fits in standard 16GB VRAM (T4 / L4)
# 'Wan-AI/Wan2.1-T2V-14B-Diffusers'  -> Requires A100 / H100
MODEL_ID = 'Wan-AI/Wan2.1-T2V-1.3B-Diffusers'

print(f'Loading {MODEL_ID} with bfloat16 and model CPU offload...')
pipe = WanPipeline.from_pretrained(
    MODEL_ID,
    torch_dtype=torch.bfloat16
)
pipe.enable_model_cpu_offload()
pipe.enable_vae_slicing()
print('Wan 2.1 Video Pipeline Ready!')
```

```
# @title 3. Generate SOTA Video with Dynamic Motion (Brain + Director)
prompt = (
    'Goddess Durga performing a divine sacred war dance (Rudra Tandava), '
    'ten radiant arms holding glowing golden trishula and sword cutting trails of celestial light, '
    'flowing crimson and golden silk saree, majestic roaring lion beside her, '
    'dramatic volumetric lighting, cinematic 8k masterpiece, realistic human face and graceful body motion, '
    'cosmic embers and floating lotus petals.'
)
negative_prompt = 'static, blurry, deformed limbs, distorted face, low quality, artifacts, watermark'

print('Generating video (81 frames / ~5 seconds at 16fps)... Please wait...')
output = pipe(
    prompt=prompt,
    negative_prompt=negative_prompt,
    height=480,
    width=832,
    num_frames=81,
    guidance_scale=6.0,
    num_inference_steps=30
).frames[0]

output_path = 'durga_wan2_1_generated.mp4'
export_to_video(output, output_path, fps=16)
print(f'Video successfully generated at: {output_path}')
```

```
# @title 4. Preview and Download the Generated Video
from IPython.display import HTML
from base64 import b64encode
from google.colab import files

mp4 = open('durga_wan2_1_generated.mp4', 'rb').read()
data_url = 'data:video/mp4;base64,' + b64encode(mp4).decode()
display(HTML(f'<video width=640 controls autoplay loop><source src="{data_url}" type="video/mp4"></video>'))

# Uncomment below to download directly to your PC:
# files.download('durga_wan2_1_generated.mp4')
```
