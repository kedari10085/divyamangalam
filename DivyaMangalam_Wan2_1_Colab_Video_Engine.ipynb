```
# @title 2. Select AI Video Diffusion Model from Dropdown { run: "auto" }
selected_video_model = "Wan-AI/Wan2.1-T2V-1.3B-Diffusers (Lightweight - Free Colab T4)"  # @param ["Wan-AI/Wan2.1-T2V-1.3B-Diffusers (Lightweight - Free Colab T4)", "Wan-AI/Wan2.1-T2V-14B-Diffusers (Studio Quality 28GB - Colab A100/L4)", "Lightricks/LTX-Video (0.95B Ultra-Fast - Any GPU)", "THUDM/CogVideoX-2b (Fast 2B - Free T4)", "THUDM/CogVideoX-5b (High Fidelity 5B)"]

# Map dropdown selection to Hugging Face Model ID
MODEL_MAP = {
    "Wan-AI/Wan2.1-T2V-1.3B-Diffusers (Lightweight - Free Colab T4)": {
        "id": "Wan-AI/Wan2.1-T2V-1.3B-Diffusers",
        "type": "wan"
    },
    "Wan-AI/Wan2.1-T2V-14B-Diffusers (Studio Quality 28GB - Colab A100/L4)": {
        "id": "Wan-AI/Wan2.1-T2V-14B-Diffusers",
        "type": "wan"
    },
    "Lightricks/LTX-Video (0.95B Ultra-Fast - Any GPU)": {
        "id": "Lightricks/LTX-Video",
        "type": "ltx"
    },
    "THUDM/CogVideoX-2b (Fast 2B - Free T4)": {
        "id": "THUDM/CogVideoX-2b",
        "type": "cogvideo"
    },
    "THUDM/CogVideoX-5b (High Fidelity 5B)": {
        "id": "THUDM/CogVideoX-5b",
        "type": "cogvideo"
    }
}

model_info = MODEL_MAP[selected_video_model]
print(f"Selected Video Model: {model_info['id']} ({model_info['type'].upper()})")

import torch
from diffusers.utils import export_to_video

if model_info['type'] == 'wan':
    from diffusers import WanPipeline
    print(f"Loading {model_info['id']} with bfloat16 and model CPU offload...")
    pipe = WanPipeline.from_pretrained(model_info['id'], torch_dtype=torch.bfloat16)
    pipe.enable_model_cpu_offload()
    pipe.enable_vae_slicing()
elif model_info['type'] == 'ltx':
    from diffusers import LTXPipeline
    print(f"Loading {model_info['id']}...")
    pipe = LTXPipeline.from_pretrained(model_info['id'], torch_dtype=torch.bfloat16)
    pipe.enable_model_cpu_offload()
elif model_info['type'] == 'cogvideo':
    from diffusers import CogVideoXPipeline
    print(f"Loading {model_info['id']}...")
    pipe = CogVideoXPipeline.from_pretrained(model_info['id'], torch_dtype=torch.bfloat16)
    pipe.enable_model_cpu_offload()

print("🎉 Video Pipeline Successfully Loaded & Ready for Generation!")
```

```
# @title 3. Select AI Brain / Storyboarder LLM from Dropdown { run: "auto" }
selected_llm = "Qwen/Qwen2.5-7B-Instruct (Best Multilingual Sanatana Epics)"  # @param ["Qwen/Qwen2.5-7B-Instruct (Best Multilingual Sanatana Epics)", "deepseek-ai/DeepSeek-R1-Distill-Qwen-8B (Reasoning & Storyboard)", "meta-llama/Llama-3.2-3B-Instruct (Lightweight)", "Local-Ollama-PC (llama3.2:3b / qwen2.5:3b)", "Moonshot-Kimi-K3.1-Pro (API Mode)"]

print(f"Active Storyboard Brain: {selected_llm}")
print("LLM Brain is configured to design rich cinematic video prompts, choreography, and camera directions.")
```

```
# @title 4. Choose Preset Sacred Theme or Enter Custom Prompt { run: "auto" }
preset_theme = "Lord Ayyappa & Lord Murugan Celestial Summit"  # @param ["Lord Ayyappa & Lord Murugan Celestial Summit", "Prince Manikandan Fearlessly Riding Royal Bengal Tigress", "Lord Venkateswara Suprabhata Seva Garbhagriha", "Goddess Durga Rudra Tandava Cosmic War Dance", "Lord Shiva Nataraja Ananda Tandava", "Lord Hanuman Flying with Sanjeevani Mountain", "Custom Sacred Prompt"]

THEME_PROMPTS = {
    "Lord Ayyappa & Lord Murugan Celestial Summit": (
        "Lord Ayyappa and Lord Murugan standing together on a sacred Himalayan mountain summit at golden hour sunset, "
        "Lord Murugan holding glowing golden Vel spear with iridescent peacock beside him, "
        "Lord Ayyappa holding golden bow with majestic royal Bengal tiger sitting calmly, "
        "classical oil painting aesthetic, rich chiaroscuro, volumetric golden lighting, highly realistic human facial features."
    ),
    "Prince Manikandan Fearlessly Riding Royal Bengal Tigress": (
        "12-year-old Prince Manikandan (Lord Ayyappa) with radiant golden crown fearlessly riding a colossal powerful Royal Bengal Tigress, "
        "moving gracefully through dense misty tropical Western Ghats jungle, playful tiger cubs walking alongside, "
        "shafts of morning sunlight piercing through mossy banyan trees, classical fine-art oil painting realism."
    ),
    "Lord Venkateswara Suprabhata Seva Garbhagriha": (
        "Lord Sri Venkateswara in the golden sanctum sanctorum of Tirumala during holy morning Suprabhata Seva, "
        "radiant diamond Kiritam, white and red Urdhva Pundra Namam, Goddess Vyuha Lakshmi on chest, "
        "archakas waving 27-flame Nakshatra Maha Harathi in circular arcs, warm glowing oil lamps, falling Tulasi leaves."
    ),
    "Goddess Durga Rudra Tandava Cosmic War Dance": (
        "Goddess Durga performing a divine sacred war dance (Rudra Tandava), "
        "ten radiant arms holding glowing golden trishula and sword cutting trails of celestial light, "
        "flowing crimson and golden silk saree, majestic roaring lion beside her, volumetric lighting, cosmic embers."
    ),
    "Lord Shiva Nataraja Cosmic Dance": (
        "Lord Shiva as Nataraja performing the cosmic dance within a circular halo of sacred fire (Prabhavali), "
        "flowing matted locks, Damaru drum and fire in hands, crescent moon, holy river Ganga flowing from locks, starry cosmic void."
    ),
    "Lord Hanuman Flying with Sanjeevani Mountain": (
        "Lord Hanuman flying majestically across the starry midnight sky carrying the glowing golden Dronagiri Sanjeevani herb mountain in his palm, "
        "golden gada mace at hip, compassionate determined humanoid face, clouds parting with golden radiance."
    )
}

custom_prompt = "" # @param {type:"string"}

if preset_theme == "Custom Sacred Prompt" and custom_prompt.strip():
    active_prompt = custom_prompt.strip()
else:
    active_prompt = THEME_PROMPTS.get(preset_theme, THEME_PROMPTS["Lord Ayyappa & Lord Murugan Celestial Summit"])

negative_prompt = "static, blurry, deformed limbs, distorted face, low quality, cartoon, artifacts, watermark"

# Video generation parameters
video_duration_seconds = 5  # @param [3, 5, 8] {type:"raw"}
fps = 16
total_frames = video_duration_seconds * fps

print("🎬 Prompt:", active_prompt)
print(f"Rendering {total_frames} frames (~{video_duration_seconds}s @ {fps}fps)... Please wait...")

# Run Diffusion Pipeline
with torch.inference_mode():
    output = pipe(
        prompt=active_prompt,
        negative_prompt=negative_prompt,
        height=480,
        width=832,
        num_frames=total_frames,
        guidance_scale=6.0,
        num_inference_steps=30
    ).frames[0]

output_path = "divyamangalam_ai_generated_video.mp4"
export_to_video(output, output_path, fps=fps)
print(f"🎉 Video successfully generated at: {output_path}")
```
