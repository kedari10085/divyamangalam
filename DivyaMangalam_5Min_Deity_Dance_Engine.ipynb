## 🎵 Step 2: Download Pure Sacred Music & Art Assets

```python
import urllib.request, os
os.makedirs('audio', exist_ok=True)
os.makedirs('images', exist_ok=True)

# Download Sacred Shiva Tandava (6.5 mins)
shiva_music_url = 'https://archive.org/download/ShivaTandavaStotram_201707/Shiva_Tandava_Stotram.mp3'
urllib.request.urlretrieve(shiva_music_url, 'audio/shiva_tandava.mp3')

# Download Sacred Aditya Hrudayam (8.3 mins)
surya_music_url = 'https://archive.org/download/AdityaHridayamMantra/Aditya%20Hrudayam.mp3'
urllib.request.urlretrieve(surya_music_url, 'audio/aditya_hrudayam.mp3')

# Download High-Res Deity Artwork
urllib.request.urlretrieve('https://raw.githubusercontent.com/kedari10085/divyamangalam/main/images/shiva_kailash.jpg', 'images/shiva_kailash.jpg')
urllib.request.urlretrieve('https://raw.githubusercontent.com/kedari10085/divyamangalam/main/images/surya_bhagavan.jpg', 'images/surya_bhagavan.jpg')

print('Pure Sacred Music and Artworks downloaded!')
```

## 🎬 Step 3: Configure Your Video (60s Shorts or 5 Minutes)

```python
# @title Choose Deity and Duration
DEITY = 'shiva' # @param ['shiva', 'surya']
FORMAT = 'shorts (1080x1920)' # @param ['shorts (1080x1920)', 'full (1920x1080)']
DURATION_MINUTES = 5.0 # @param [1.0, 3.0, 5.0]

mode = 'shorts' if 'shorts' in FORMAT else 'full'
duration_sec = DURATION_MINUTES * 60.0
print(f'Ready to render: {DEITY.upper()} | {FORMAT} | {duration_sec} seconds')
```

## ⚡ Step 4: Render Pure Music & Cosmic Dance Video

```python
!curl -s -O https://raw.githubusercontent.com/kedari10085/divyamangalam/main/generate_pure_music_dance_video.py

output_name = f'{DEITY}_dance_{int(DURATION_MINUTES)}min.mp4'
print(f'Rendering {output_name} on Colab GPU...')
!python generate_pure_music_dance_video.py --deity {DEITY} --mode {mode} --duration {duration_sec} --out {output_name}
print('Video generation complete!')
```

## 💾 Step 5: Download Video to Your PC / Phone

```python
from google.colab import files
files.download(output_name)
print(f'Downloading {output_name} to your device!')
```
